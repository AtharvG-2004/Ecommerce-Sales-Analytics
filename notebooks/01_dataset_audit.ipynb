## 2. Missing Value Audit

```python
for name, df in datasets.items():
    print("=" * 70)
    print(f"TABLE: {name}")

    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df)) * 100

    missing_report = pd.DataFrame({
        "Missing_Count": missing_count,
        "Missing_Percentage": missing_percentage.round(2)
    })

    print(missing_report)
    print()
```

## 3. Duplicate Record Audit

```python
for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    
    print("=" * 70)
    print(f"TABLE: {name}")
    print(f"Duplicate rows: {duplicate_count}")
    print()
```

Primary Key Verification

```python
primary_key_candidates = {
    "customers": ["customer_id", "customer_unique_id"],
    "geolocation": ["geolocation_zip_code_prefix"],
    "order_items": ["order_id", "order_item_id"],
    "order_payments": ["order_id", "payment_sequential"],
    "order_reviews": ["review_id"],
    "orders": ["order_id"],
    "products": ["product_id"],
    "sellers": ["seller_id"],
    "product_category_name_translation": ["product_category_name"]
}
```

```python
for table_name, columns in primary_key_candidates.items():
    df = datasets[table_name]

    print("=" * 70)
    print(f"TABLE: {table_name}")

    for column in columns:
        unique_count = df[column].nunique()
        total_count = len(df)
        duplicate_count = total_count - unique_count

        print(
            f"{column}: "
            f"unique={unique_count:,}, "
            f"rows={total_count:,}, "
            f"duplicate_values={duplicate_count:,}"
        )

    print()
```

```python
# orders → customers
missing_customers = ~orders["customer_id"].isin(customers["customer_id"])

print("Orders with customer_id not found in customers:")
print(missing_customers.sum())


# order_items → orders
missing_orders_from_items = ~order_items["order_id"].isin(orders["order_id"])

print("\nOrder items with order_id not found in orders:")
print(missing_orders_from_items.sum())


# order_items → products
missing_products = ~order_items["product_id"].isin(products["product_id"])

print("\nOrder items with product_id not found in products:")
print(missing_products.sum())


# order_items → sellers
missing_sellers = ~order_items["seller_id"].isin(sellers["seller_id"])

print("\nOrder items with seller_id not found in sellers:")
print(missing_sellers.sum())


# order_payments → orders
missing_orders_from_payments = ~order_payments["order_id"].isin(orders["order_id"])

print("\nPayments with order_id not found in orders:")
print(missing_orders_from_payments.sum())


# order_reviews → orders
missing_orders_from_reviews = ~order_reviews["order_id"].isin(orders["order_id"])

print("\nReviews with order_id not found in orders:")
print(missing_orders_from_reviews.sum())
```

```python
# Check how many ID values contain quotation marks

for name, df in datasets.items():
    print("=" * 70)
    print(f"TABLE: {name}")

    for column in df.columns:
        if df[column].dtype == "object" or str(df[column].dtype) == "str":
            quoted_count = df[column].astype("string").str.contains(
                '"',
                na=False
            ).sum()

            if quoted_count > 0:
                print(f"{column}: {quoted_count:,} values contain quotation marks")

    print()
```

```python
def normalize_id(series):
    return (
        series.astype("string")
        .str.strip()
        .str.strip('"')
    )


# Orders → Customers
orders_customer_normalized = normalize_id(orders["customer_id"])
customers_id_normalized = normalize_id(customers["customer_id"])

print(
    "Orders → Customers unmatched after ID normalization:",
    (~orders_customer_normalized.isin(customers_id_normalized)).sum()
)


# Order Items → Sellers
items_seller_normalized = normalize_id(order_items["seller_id"])
sellers_id_normalized = normalize_id(sellers["seller_id"])

print(
    "Order Items → Sellers unmatched after ID normalization:",
    (~items_seller_normalized.isin(sellers_id_normalized)).sum()
)


# Order Items → Orders
items_order_normalized = normalize_id(order_items["order_id"])
orders_id_normalized = normalize_id(orders["order_id"])

print(
    "Order Items → Orders unmatched after ID normalization:",
    (~items_order_normalized.isin(orders_id_normalized)).sum()
)


# Payments → Orders
payments_order_normalized = normalize_id(order_payments["order_id"])

print(
    "Payments → Orders unmatched after ID normalization:",
    (~payments_order_normalized.isin(orders_id_normalized)).sum()
)


# Reviews → Orders
reviews_order_normalized = normalize_id(order_reviews["order_id"])

print(
    "Reviews → Orders unmatched after ID normalization:",
    (~reviews_order_normalized.isin(orders_id_normalized)).sum()
)
```
